```python
def print_numbers():
    for number in range(1,6):
            print(number)
    
print_numbers()
            
```

```python
def print_even():
    for number in range(1,11):
        if number%2==0:
            print(number)

print_even()
```

```python
def sum_numbers():
    total=0
    
    for number in range(1,6):
        total=total+number

        
    return total
print(sum_numbers())
```
