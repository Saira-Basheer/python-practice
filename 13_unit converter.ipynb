```python
choice = input("choice conversion(km_to_m,c_to_f,kg_to_g:")
value=float(input("enter value:"))

if choice=="km_to_m":
    result=value*1000
    print("meters:" , result)

elif choice=="c_to_f":
    result=(value*9/5)+32
    print("Fahrenheit:" , result)
    
elif choice=="kg_to_g":
    result=value*1000
    print("grams:" , result)
    
else:
    print("invalid ")

```
